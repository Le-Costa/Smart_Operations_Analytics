```python
import pandas as pd
path='C:\\Users\\nadia\\OneDrive\\Desktop\\Leticia\\Documentos\\Projetos\\Smart_Operations_Analytics'

products = pd.read_csv(path + '\\dim_products.csv')
customers = pd.read_csv(path + '\\dim_customers.csv')    
suppliers = pd.read_csv(path + '\\dim_suppliers.csv')   
inventory = pd.read_csv(path + '\\fact_inventory_monthly.csv')     
orders = pd.read_csv(path + '\\fact_orders.csv')
production = pd.read_csv(path + '\\fact_production.csv')
```

# Análise exploratória

```python
# função para poder visualizar as informações de forma enxuta

# %% Initial Data Understanding

def data_overview(df, name):

    print(f"DATASET: {name}")
    print(f"{'=' * 40}")

    print(f"Rows and Columns: {df.shape}")

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isna().sum())

    print("\nDuplicated Rows:")
    print(df.duplicated().sum())

    print("\nStatistical Summary:")
    print(df.describe())


# %% Overview

data_overview(products, "Products")
data_overview(customers, "Customers")
data_overview(suppliers, "Suppliers")

data_overview(production, "Production")
data_overview(inventory, "Inventory")
data_overview(orders, "Orders")
```

As tabelas dimensionais não vão precisar de muito tratamento. Vai ser necessário maior atenção na tabela de ordens.
